```python
def make_child_id(cid: tuple, child_number: int) -> tuple:
    return cid + (child_number,)


def mutated_parameters(clone: dict, fitness_gain: float) -> dict:
    return {
        'lambda': clone['lambda'] * (1.0 + fitness_gain),
        'mu': clone['mu'],
        'nu': clone['nu'],
    }
```

```python
def mutate_clone(clones,cid):
    clone=clones[cid]
    clone["N"] -= 1

    # Create new child clone
    clone["children_count"] += 1
    child_number = clone["children_count"]
    child_id = make_child_id(cid, child_number)

    params = mutated_parameters(clone, fitness_gain=fitness_gain)

    clones[child_id] = {
        "N": 1,
        "lambda": params["lambda"],
        "mu": params["mu"],
        "nu": params["nu"],
        "parent": cid,
        "children_count": 0,
    }
```
