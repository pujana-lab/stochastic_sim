Creamos un objeto básico que tenga la función de clonar y matar, y lo implementamos

```python
class Population:
    def __init__(self, N):
        self.N = N

    def divide(self):
        self.N += 1
        
    def kill(self):
        self.N -= 1

    def __str__(self):
        return f"Population size: {self.N}"

```

```python
pop = Population(100)
print(pop)
pop.divide()
print(pop)
pop.kill()
print(pop)
```
