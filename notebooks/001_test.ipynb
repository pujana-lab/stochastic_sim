Notebook 1 de ejemplo. Vamos a crear un objeto básico que tenga poblaciones y que permita trabajar sobre ellas. En esta primera fase vamos a crear una base que clona y divide


```python
def divide(clones, cid):
    clone=clones[cid]
    clone["N"]+= 1
    return clones
    
def kill(clones,cid):
    clone=clones[cid]
    clone["N"]-=1
    return clones
```

```python
# Creamos una población vacía mínima
clones = {(1):{'N': 100}}
```

```python
clones2 = divide(clones,1 )

print(clones2)
```

```python
clones3 = kill(clones,1 )
clones3
```
