```python
from typing import  Tuple

CloneId = Tuple[int, ...]

#def make_child_id(parent_id: tuple, child_number: int)-> CloneId:
#   return parent_id + (child_number,)

class ChildLabelBuilder:
    def generate(self, parent_id: CloneId, child_number: int) -> CloneId:
        return parent_id + (child_number,)

class Population:
    def __init__(self, N, id = ()):
        self.N = N
        self.id = id

    def divide(self):
        self.N += 1
        
    def kill(self):
        self.N -= 1

    def mutate(self, child_number):
        self.kill()
        child_id = ChildLabelBuilder().generate(self.id, child_number)
        return Population(1, child_id)



    def __str__(self):
        return f"Population {self.id} size: {self.N}"


```

```python
pop = Population(100)
print(pop)
print('-----------')
pop2 = pop.mutate(1)
print(pop)
print(pop2)
```
