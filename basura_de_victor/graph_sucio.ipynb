```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1=pd.read_csv('history.csv')
df2 = pd.read_csv('hist2.csv')
```

## using old carrying capacity limiter

```python
df=df1[df1['time']<200]
del df1
fig,ax=plt.subplots(figsize=(10,6))
plt.plot(df['time'],df['N'],df['time'],df['rb'],df['time'],df['rd'])    
plt.show()
```

## Using new version

```python
df=df2[df2['time']<200]
del df2
fig,ax=plt.subplots(figsize=(10,6))
plt.plot(df['time'],df['N'],df['time'],df['rb'],df['time'],df['rd'])    
plt.show()
del df
```

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt


outdir = Path("figures")
outdir.mkdir(exist_ok=True)

t = np.linspace(0, 200, 400)

# -----------------------------
# Placeholder 1: before change
# Original birth-limited crowding:
# equilibrium sits below nominal K
# -----------------------------
H_before = 60
N0 = 10
k_rate = 0.05

before = H_before - (H_before - N0) * np.exp(-k_rate * t)

plt.figure(figsize=(7, 4.2))
plt.plot(t, before, label="WT population")
plt.axhline(100, linestyle="--", label="Nominal K")
plt.axhline(H_before, linestyle=":", label="Observed homeostasis")
plt.xlabel("Time")
plt.ylabel("Population size")
plt.title("Placeholder simulation: original birth-limited crowding")
plt.legend()
plt.tight_layout()
plt.savefig(outdir / "before_change_placeholder.png", dpi=200)
plt.show()


# -----------------------------
# Placeholder 2: after change
# Calibrated crowding:
# homeostasis recovered at H=100
# -----------------------------
H_after = 100

after = H_after - (H_after - N0) * np.exp(-k_rate * t)

plt.figure(figsize=(7, 4.2))
plt.plot(t, after, label="WT population")
plt.axhline(H_after, linestyle="--", label="Target homeostasis H")
plt.xlabel("Time")
plt.ylabel("Population size")
plt.title("Placeholder simulation: calibrated homeostatic crowding")
plt.legend()
plt.tight_layout()
plt.savefig(outdir / "after_change_placeholder.png", dpi=200)
plt.show()
```
